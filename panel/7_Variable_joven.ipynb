# Variable de edad aproximada

La cedula colombiana se asigna por rangos segun el ano de nacimiento, asi
que el numero de digitos sirve como aproximacion: diez digitos o mas
corresponde a las series mas recientes. La variable se agrega a la base de
historial.

```python
import pandas as pd

ruta_archivo = r"C:\Users\DiDi\Downloads\base_historial_escalonado.parquet"

df = pd.read_parquet(ruta_archivo)

muestra = df

muestra['joven'] = (muestra['IDENTIFICACION'].astype(str).str.len() >= 10).astype(int)

print(muestra[['IDENTIFICACION', 'joven']].head())

ruta_salida = r"C:\Users\DiDi\Downloads\base_historial_escalonado.parquet"
muestra.to_parquet(ruta_salida)

print("¡Muestra extraída y guardada con éxito!")
```
